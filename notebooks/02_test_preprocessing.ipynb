```python
print("Pemain target:")
display(
    model.df.loc[
        model.df["player_id"] == 1,
        ["player_id", "player_name", "profile_text"]
    ]
)

print("Profil Albert Streit:")
display(
    model.df.loc[
        model.df["player_id"] == 1080,
        ["player_id", "player_name", "profile_text"]
    ]
)
```

```python
silvio = model.df.loc[
    model.df["player_id"] == 1, "profile_text"
].iloc[0]

albert = model.df.loc[
    model.df["player_id"] == 1080, "profile_text"
].iloc[0]

print("Profil Silvio Adzic:")
print(silvio)

print("\nProfil Albert Streit:")
print(albert)

print("\nApakah profilnya identik?", silvio == albert)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

idx_silvio = model.df.index[
    model.df["player_id"] == 1
][0]

idx_albert = model.df.index[
    model.df["player_id"] == 1080
][0]

skor = cosine_similarity(
    model.tfidf_matrix[idx_silvio],
    model.tfidf_matrix[idx_albert]
)[0, 0]

print("Skor kemiripan Silvio dan Albert:", round(skor, 6))
```
